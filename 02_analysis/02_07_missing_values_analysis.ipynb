### 07. Анализ пропусков как скрытого фактора

Студенты, не указавшие уровень образования родителей, отличаются от остальных? Может ли сам факт пропуска (нежелание / невозможность ответить) коррелировать с их успеваемостью, ментальным здоровьем или социальным статусом?

```python
import sys
import os
from pathlib import Path

project_root = Path(os.getcwd()).parent 
sys.path.insert(0, str(project_root))

from src.preparation import load_df
df = load_df()
```

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import src.constants as cnst 

plt.rcParams['font.family'] = 'Arial'

df_02_07 = df.copy()

df_02_07['is_parent_edu_hidden'] = df_02_07.parental_education_level.isna()

res_02_07 = df_02_07.groupby('is_parent_edu_hidden').agg(
    avg_exam_score=('exam_score', 'mean'), 
    avg_study_hours=('study_hours_per_day', 'mean'), 
    pct_part_time_job=('part_time_job', lambda i: (i == 'Yes').sum() / i.count() * 100), 
    avg_attendance_pct=('attendance_percentage', 'mean'), 
    pct_poor_internet_quality=('internet_quality', lambda i: (i == 'Poor').sum() / i.count() * 100), 
    avg_mental_health=('mental_health_rating', 'mean')
).round(1)

display(res_02_07)

df_02_07_flt = df_02_07.loc[df_02_07.is_parent_edu_hidden == True]
pd.crosstab(
    index=df_02_07_flt.age, columns=df_02_07_flt.gender
)
```

#### 07. Выводы 

Анализ пропусков показал, что они [пропуски] распределены равномерно. Студенты, скрывшие уровень образования родителей, не имеют значимых отличий по успеваемости, ментальному здоровью, посещаемости, полу, возрасту. Кросстабуляция не выявила специфических «скрытных» демографических групп. Таким образом, пропуски в столбце parental_education_level являются случайными (MCAR — Missing Completely At Random) и не влияют на анализ.
