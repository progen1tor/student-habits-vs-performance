### 0. Первичный осмотр данных 

```python
import pandas as pd 

df = pd.read_csv('../data/raw.csv')
df.info()
display(df.head())
display(df.describe())
display(df.select_dtypes(include='str').describe())
```

### 1. Проверка корректности данных 

```python
df.isna().sum()  # -> только 91 пропуск в колонке `parental_education_level`
df.duplicated().sum()  # 0 -> полных дубликатов нет 
df.student_id.nunique() == len(df)  # True -> ID студентов уникальны 
```

### 2. Пояснения к колокнкам
***student_id*** — уникальный идентификатор студента (строка)

***age*** — возраст студента в годах (целое число, 17-24)

***gender*** — пол студента (строковая категория, Male / Female / Other)

***study_hours_per_day*** — сколько часов в день студент тратит на учёбу (дробное число, 0-8.3)

***social_media_hours*** — сколько часов в день уходит на соцсети (дробное число, 0-7.2)

***netflix_hours*** — сколько часов в день студент смотрит Netflix (дробное число, 0-5.4)

***part_time_job*** — наличие подработки (Yes / No)

***attendance_percentage*** — процент посещаемости занятий (дробное число, 56-100%)

***sleep_hours*** — сколько часов в сутки студент спит (дробное число, 3.2-10)

***diet_quality*** — качество питания (строковая категория, Poor / Fair / Good)

***exercise_frequency*** — частота физических упражнений (целое число, 0-6)

***parental_education_level*** — уровень образования родителей (строковая категория, High School / Bachelor / Master), содержит 91 пропуск  

***internet_quality*** — качество интернет-соединения (строковая категория, Good / Average / Poor)

***mental_health_rating*** — оценка психического здоровья (целое число, 1-10)

***extracurricular_participation*** — участие во внеучебной деятельности (Yes / No)

***exam_score*** — итоговый балл за экзамен (дробное число, 18.4-100)
